```python
df['term'] = df['term'].astype(str).str.extract('(\d+)').astype(float)
```

```python
def clean_emp_length(x):
    if pd.isna(x): return 0
    x = str(x)
    if '<' in x: return 0
    if '+' in x: return 10
    # extract digits
    digits = ''.join(filter(str.isdigit, x))
    return int(digits) if digits else 0

if 'emp_length' in df.columns:
    df['emp_length'] = df['emp_length'].apply(clean_emp_length)
```

```python
df['earliest_cr_line'] = pd.to_datetime(df['earliest_cr_line'], format='%b-%y', errors='coerce')
# Use 2025 as current reference year
df['credit_history_years'] = 2025 - df['earliest_cr_line'].dt.year
# Fill missing dates with median (rare case)
df['credit_history_years'] = df['credit_history_years'].fillna(df['credit_history_years'].median())
# Drop original
df = df.drop(columns=['earliest_cr_line'])
```

```python
df['has_delinq_history'] = np.where(df['mths_since_last_delinq'].isnull(), 0, 1)
# Step D2: Fill with 1000 (representing "Infinity" / Never)
df['mths_since_last_delinq'] = df['mths_since_last_delinq'].fillna(1000)
```

```python
df['mths_since_last_major_derog'] = df['mths_since_last_major_derog'].fillna(1000)
```
